```python
# Stacking BackScattering Energy
import pandas as pd
import matplotlib.pyplot as plt

# Load GaNe3/Backscatter_energe3.txt (assume whitespace or tab delimited, two columns)
bs_e_gane3 = pd.read_csv('GaNe3/Backscattered_Energy_e3.txt', sep='\\t|\\s+', engine='python', header=None)
if bs_e_gane3.shape[1] == 2:
    bs_e_gane3.columns = ['Energy', 'Counts']

print(bs_e_gane3.head())

plt.plot(bs_e_gane3['Energy'], bs_e_gane3['Counts'], label='Backscatter Energy GaNe3')
plt.xlabel('Energy (keV)')
plt.ylabel('Counts')
plt.title('Backscattered Energy Spectrum (GaNe3)')
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# Stacking BackScattering Energy
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Load 30KeV data
bs_e_30 = pd.read_csv('30KeV/BackScattered_Energy_30KeV', sep='\t', header=1)

# Fit a 3rd order polynomial to the data
x = bs_e_30['Energy (KeV)'].values
y = bs_e_30['Hits (Normalized)'].values
# coeffs_3 = np.polyfit(x, y, deg=3)
# coeffs_4 = np.polyfit(x, y, deg=4)  # Example for 4th order polynomial if needed
coeffs_5 = np.polyfit(x, y, deg=5)  # Example for 5th order polynomial if needed

# fit_y_3 = np.polyval(coeffs_3, x)
# fit_y_4 = np.polyval(coeffs_4, x)
fit_y_5 = np.polyval(coeffs_5, x)

# # Print coefficients
# print('3rd order polynomial coefficients:', coeffs_3)
# print('4th order polynomial coefficients:', coeffs_4)
print('5th order polynomial coefficients:', coeffs_5)

# Plotting
plt.plot(x, y, label='Backscattered Energy 30KeV')
# plt.plot(x, fit_y_3, label='3rd Order Polynomial Fit', linestyle='--', color='red')
# plt.plot(x, fit_y_4, label='4th Order Polynomial Fit', linestyle='--', color='green')
plt.plot(x, fit_y_5, label='5th Order Polynomial Fit', linestyle='--', color='red')

plt.xlabel('Energy (keV)')
plt.ylabel('Hits (Normalized)')
plt.title('CASINO Backscattered Energy Spectrum with 3rd Order Fit')
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()
```
