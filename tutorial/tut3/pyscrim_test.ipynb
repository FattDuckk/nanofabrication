```python
from srim import Ion, Layer, Target, TRIM, Element

# 1. Define ion (Oxygen, 2 eV)
ion = Ion('O', energy=2.0)  # 2 eV Oxygen

# 2. Define target layer (Diamond: Carbon, width=10000 Å, density=3.53 g/cm³)
# Element expects: symbol
carbon = Element('C')
layer = Layer({carbon: 1.0}, density=3.53, width=10000.0)  # width in Ångströms

# 3. Combine into target
target = Target([layer])

# 4. Run TRIM
trim = TRIM(ion, target, number_ions=1000)
trim.run()  # Will call SRIM if installed
```
