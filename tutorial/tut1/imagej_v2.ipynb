```python
import imagej
# 'interactive' → shows the ImageJ UI
# 'headless' → no UI, good for batch processing
ij = imagej.init('sc.fiji:fiji', mode='interactive')
# ij.ui().showUI()  # optional: only does something in interactive mode
```

```python
# Change this to your file path
image_path = 'balls1-2048x2048.tif'

img_java = ij.io().open(image_path)                 # Java ImgPlus
ij.ui().show("Input", img_java)               # Show in ImageJ UI (optional for visual check)
img = ij.py.from_java(img_java)               # Convert to NumPy/xarray for inline display

# Inline display in notebook (matplotlib)
import matplotlib.pyplot as plt
plt.imshow(img if img.ndim==2 else img[...,0], cmap='gray')
plt.title("Input image")
plt.axis('off')
plt.show()
```
