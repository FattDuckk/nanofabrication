```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the backscattered energy data (tab or whitespace delimited, header on second line)
df = pd.read_csv('BackScattered_Energy_30KeV', sep='	', header=1)

# Check the first few rows
print(df.head())

# Plotting
plt.plot(df['Energy (KeV)'], df['Hits (Normalized)'], label='Backscattered Energy')
plt.xlabel('Energy (keV)')
plt.ylabel('Hits (Normalized)')
plt.title('CASINO Backscattered Energy Spectrum')
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# Stacking BackScattering Energy
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Load 30KeV data
bs_e_30 = pd.read_csv('30KeV/BackScattered_Energy_30KeV', sep='\t', header=1)

# Fit a 3rd order polynomial to the data
x = bs_e_30['Energy (KeV)'].values
y = bs_e_30['Hits (Normalized)'].values
coeffs = np.polyfit(x, y, deg=3)
fit_y = np.polyval(coeffs, x)

# Print coefficients
print('3rd order polynomial coefficients:', coeffs)

# Plotting
plt.plot(x, y, label='Backscattered Energy 30KeV')
plt.plot(x, fit_y, label='3rd Order Polynomial Fit', linestyle='--', color='red')
plt.xlabel('Energy (keV)')
plt.ylabel('Hits (Normalized)')
plt.title('CASINO Backscattered Energy Spectrum with 3rd Order Fit')
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()
```
