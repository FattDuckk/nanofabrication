# CL Spectra Analysis

Simple visualization of 30 CL spectra from CL_data.csv

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Load data
df = pd.read_csv('CL_data.csv', skiprows=[1])
energy = pd.to_numeric(df.iloc[:, 0]).values
spectra = df.iloc[:, 1:31].apply(pd.to_numeric).values

print(f"Loaded {spectra.shape[1]} spectra, energy range: {energy[0]:.2f}-{energy[-1]:.2f} eV")
```

## Figure 1: All 30 Spectra

```python
fig, ax = plt.subplots(figsize=(12, 8))

# Use spectrum colormap - progression from purple to red
colors = plt.cm.plasma(np.linspace(0, 1, 30))

for i in range(30):
    ax.plot(energy, spectra[:, i], color=colors[i], alpha=0.8, linewidth=1.5)

ax.set_xlabel('Energy (eV)')
ax.set_ylabel('CL Intensity (a.u.)')
ax.set_title('All 30 CL Spectra')
ax.grid(True, alpha=0.3)

# Add colorbar to show spectrum progression - FIX: specify ax
sm = plt.cm.ScalarMappable(cmap=plt.cm.plasma, norm=plt.Normalize(vmin=1, vmax=30))
sm.set_array([])
cbar = plt.colorbar(sm, ax=ax)
cbar.set_label('Spectrum Number')

plt.tight_layout()
plt.show()
```

## Figure 2: Spectra 1-10

```python
fig, ax = plt.subplots(figsize=(12, 8))

# Use spectrum colors but spread them out more for better distinction
# Use the full range of plasma for just 10 spectra to maximize difference
colors = plt.cm.plasma(np.linspace(0, 0.4, 10))  # First part of spectrum (purple to pink)

for i in range(10):
    ax.plot(energy, spectra[:, i], color=colors[i], linewidth=1.5, label=f'Spectrum {i+1}')

ax.set_xlabel('Energy (eV)')
ax.set_ylabel('CL Intensity (a.u.)')
ax.set_title('CL Spectra 1-10')
ax.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

## Figure 3: Spectra 11-20

```python
fig, ax = plt.subplots(figsize=(12, 8))

# Use middle section of spectrum (pink to orange)
colors = plt.cm.plasma(np.linspace(0.3, 0.7, 10))

for i in range(10, 20):
    ax.plot(energy, spectra[:, i], color=colors[i-10], linewidth=1.5, label=f'Spectrum {i+1}')

ax.set_xlabel('Energy (eV)')
ax.set_ylabel('CL Intensity (a.u.)')
ax.set_title('CL Spectra 11-20')
ax.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

## Figure 4: Spectra 21-30

```python
fig, ax = plt.subplots(figsize=(12, 8))

# Use final section of spectrum (orange to yellow)
colors = plt.cm.plasma(np.linspace(0.6, 1.0, 10))

for i in range(20, 30):
    ax.plot(energy, spectra[:, i], color=colors[i-20], linewidth=1.5, label=f'Spectrum {i+1}')

ax.set_xlabel('Energy (eV)')
ax.set_ylabel('CL Intensity (a.u.)')
ax.set_title('CL Spectra 21-30')
ax.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
