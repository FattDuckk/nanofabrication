# Depth-Resolved Cathodoluminescence Analysis of GaN-Based LED Quantum Well Structures

## Abstract

We present a comprehensive depth-resolved cathodoluminescence (CL) spectroscopy study of a multi-layered gallium nitride (GaN) light-emitting diode (LED) structure containing two quantum wells. Variable beam energy CL spectroscopy (1-30 keV) was used to probe the optical properties as a function of depth within the device structure. Five distinct emission bands were identified: near-bandgap emission (NBE, 3.385 eV), blue luminescence (BL, 2.8 eV), two quantum well emissions (QW1 at 2.634 eV, QW2 at 2.5 eV), and yellow luminescence (YL, 2.2 eV). Depth profiling analysis revealed that the quantum wells are located at different depths within the active region, with QW2 at approximately 350 nm and QW1 at approximately 720 nm below the surface, as estimated using the Kanaya-Okayama equation. The BL emission shows realistic surface-preferential behavior, while the NBE profile exhibits physically unrealistic behavior suggesting potential measurement artifacts or peak misidentification. This work demonstrates the effectiveness of variable-energy CL for characterizing the internal structure of complex semiconductor devices and identifying the spatial distribution of different emission mechanisms in LED structures.

**Keywords:** Cathodoluminescence, GaN, LED, quantum wells, depth profiling, semiconductor characterization

## 1. Introduction

Gallium nitride (GaN) and its alloys have revolutionized the field of optoelectronics, enabling the development of highly efficient blue and white light-emitting diodes (LEDs) that have transformed lighting technology worldwide [1,2]. The success of GaN-based LEDs relies on sophisticated multi-layered heterostructures that incorporate quantum wells (QWs) to enhance carrier confinement and radiative recombination efficiency [3]. However, characterizing the internal structure and optical properties of these complex devices remains challenging due to their buried active regions and nanoscale dimensions.

Cathodoluminescence (CL) spectroscopy has emerged as a powerful technique for investigating the optical properties of semiconductor materials and devices with high spatial resolution [4,5]. When combined with variable beam energy analysis, CL enables depth-resolved characterization by controlling the electron penetration depth through the accelerating voltage [6]. This approach has proven particularly valuable for studying buried interfaces, defect distributions, and the optical properties of quantum well structures in III-nitride devices [7,8].

The fundamental principle of depth-resolved CL relies on the relationship between electron beam energy and penetration depth in the material. Higher beam energies probe deeper into the sample, allowing researchers to map the depth distribution of different emission mechanisms [9]. For GaN-based structures, typical emission bands include near-bandgap emission (NBE) from the host material, quantum well emissions in the blue-green spectral region, and various defect-related emissions such as yellow luminescence (YL) [10,11].

Previous studies have demonstrated the effectiveness of variable-energy CL for characterizing GaN-based devices, including the identification of quantum well positions [12], the analysis of strain effects [13], and the investigation of defect distributions [14]. However, careful analysis is required to distinguish between realistic depth-dependent behavior and potential measurement artifacts, particularly for emissions that may overlap spectrally or suffer from experimental limitations.

In this work, we present a comprehensive depth-resolved CL analysis of a GaN-based LED structure containing two quantum wells. Our objectives are to: (1) characterize the spectral evolution of all emission bands as a function of beam energy, (2) determine the depth positions of the quantum wells using the Kanaya-Okayama penetration depth model, and (3) critically evaluate the physical realism of the observed depth profiles to identify potential measurement artifacts.

## 3. Experimental Methodology

### 3.1 Instrumentation

Cathodoluminescence measurements were performed using a scanning electron microscope (SEM) equipped with a cathodoluminescence detection system. The CL signal was collected using a parabolic mirror positioned above the sample surface, which efficiently collects light over a wide solid angle and directs it to a spectrometer for wavelength-resolved detection.

### 3.2 Measurement Procedure

Depth-resolved CL analysis was conducted by systematically varying the electron beam energy from 1 keV to 30 keV to control the penetration depth into the sample. The initial spectrum was acquired using an electron beam energy (E₀) of 1 keV and a beam current (I₀) of 20 nA. For all subsequent measurements, the beam power (E₀I₀) was maintained constant to ensure consistent excitation conditions across different penetration depths.

This constant power approach compensates for the natural decrease in generation efficiency at higher beam energies, providing more reliable quantitative comparisons between spectra acquired at different beam energies. A total of 30 spectra were collected, corresponding to beam energies from 1 keV to 30 keV in 1 keV increments.

### 4.4 Depth Profiling Analysis: Methodology and Theory

Depth profiling in cathodoluminescence involves analyzing how the intensity of specific emission bands varies with beam energy, which directly correlates with the electron penetration depth. For this GaN-based LED structure, higher beam energies penetrate deeper into the device, allowing systematic probing of different structural regions:

- **Surface layers (1-5 keV)**: p-type GaN contact and surface regions
- **Intermediate depths (5-15 keV)**: Active region containing quantum wells  
- **Deep regions (15-30 keV)**: n-type GaN layers and substrate interface

#### 4.4.1 Peak Extraction Methodology

For accurate depth profiling, several approaches can be employed to extract emission intensities:

1. **Peak Height Method (Used in this study)**: Extract the maximum intensity within a specified energy range around each emission center. This approach is effective for well-separated peaks with good signal-to-noise ratios.

2. **Peak Integration Method**: Calculate the integrated area under each emission band after background subtraction. This method is more robust for overlapping peaks and provides better quantitative accuracy.

3. **Multi-peak Fitting**: Simultaneously fit analytical functions (Gaussian, Lorentzian) to all peaks in a spectrum. This approach accounts for peak overlap and provides the most accurate intensity extraction.

For this analysis, the peak height method was selected due to the well-separated nature of most emission bands and the high signal-to-noise ratio of the data. The method involves identifying the maximum intensity within energy ranges of ±0.05-0.075 eV around each emission center.

#### 4.4.2 Physical Considerations for LED Structures

Several factors influence the interpretation of CL depth profiles in LED structures:

- **Quantum well dimensions**: QWs are typically 2-10 nm thick, much smaller than the electron interaction volume
- **Electric fields**: The p-n junction creates built-in electric fields affecting carrier transport
- **Doping variations**: Different doping levels throughout the structure affect recombination efficiency
- **Carrier diffusion**: Photo-generated carriers can diffuse before recombination, complicating depth resolution

```python
# Define emission band centers and ranges for analysis
emission_bands = {
    'NBE': {'center': 3.385, 'range': 0.1},   # Near-bandgap emission
    'BL': {'center': 2.8, 'range': 0.15},     # Blue luminescence  
    'QW1': {'center': 2.634, 'range': 0.1},   # Quantum well 1
    'QW2': {'center': 2.5, 'range': 0.1},     # Quantum well 2
    'YL': {'center': 2.2, 'range': 0.15}      # Yellow luminescence
}

def extract_peak_intensity(energy_array, spectrum, peak_energy, energy_range):
    """
    Extract peak intensity by finding maximum in a specified energy range
    This simulates peak fitting by taking the maximum value near the peak
    """
    # Find indices within the energy range around the peak
    mask = (energy_array >= peak_energy - energy_range/2) & (energy_array <= peak_energy + energy_range/2)
    
    if np.any(mask):
        # Return the maximum intensity in this range
        return np.max(spectrum[mask])
    else:
        return 0

# Extract depth profiles for each emission band
depth_profiles = {}

for band_name, band_info in emission_bands.items():
    intensities = []
    
    for i in range(30):  # For each beam energy (1-30 keV)
        spectrum = spectra[:, i]
        peak_intensity = extract_peak_intensity(energy, spectrum, 
                                              band_info['center'], 
                                              band_info['range'])
        intensities.append(peak_intensity)
    
    depth_profiles[band_name] = np.array(intensities)

print("Extracted depth profiles for all emission bands:")
for band in depth_profiles.keys():
    print(f"- {band}: Peak intensities extracted")
```

### Discussion: Peak Fitting for Accurate Depth Profiles

**Current Method**: We extracted peak intensities by finding the maximum value within a specified energy range around each emission band center. This provides a reasonable approximation but could be improved.

**Advanced Peak Fitting Approaches for Better Accuracy:**

1. **Gaussian/Lorentzian Fitting**: 
   - Fit analytical functions to each peak
   - Extract integrated area under the curve rather than just peak height
   - More robust against noise and baseline variations

2. **Multi-peak Fitting**:
   - Simultaneously fit all peaks in a spectrum
   - Account for peak overlap (especially QW1 and QW2)
   - Constrain peak positions and widths based on physical expectations

3. **Background Subtraction**:
   - Remove polynomial or exponential background
   - Use adjacent spectral regions to define baseline
   - Improves signal-to-noise ratio for weak emissions

4. **Normalization Strategies**:
   - Normalize to total integrated intensity
   - Use reference peaks that don't vary with depth
   - Account for variations in measurement conditions

**Physical Interpretation**:
The depth profiles reveal how different emission mechanisms respond to varying interaction volumes as beam energy increases. Higher beam energies probe deeper into the sample, potentially accessing different structural regions or defect distributions.

### 4.6 Quantum Well Depth Estimation

#### 4.6.1 Theoretical Background: Kanaya-Okayama Equation

The depth distribution of electron-hole pair generation in a solid can be estimated using the Kanaya-Okayama equation, which provides the maximum penetration depth of electrons:

**R = 0.0276 × A × E^1.67 / (ρ × Z^0.89)**

Where:
- R = maximum penetration depth (μm)
- A = atomic mass (g/mol)
- E = beam energy (keV)
- ρ = density (g/cm³)
- Z = atomic number

For GaN material, the relevant parameters are:
- A = 69.7 g/mol (average of Ga and N)
- ρ = 6.15 g/cm³
- Z = 31.5 (average atomic number)

#### 4.6.2 Depth Correlation Methodology

The quantum well positions can be estimated by correlating the beam energy that maximizes each QW emission with the corresponding electron penetration depth. The underlying assumption is that maximum emission intensity occurs when the electron beam optimally excites the quantum well region.

However, it is important to note that peak carrier generation typically occurs at approximately 1/3 of the maximum penetration depth, rather than at the maximum depth itself. This correction factor accounts for the depth distribution of energy deposition within the interaction volume.

```python
# Material properties for GaN (common quantum well host material)
# These values are typical for GaN-based quantum well structures
A_GaN = 69.7  # Atomic mass of GaN (average of Ga and N)
rho_GaN = 6.15  # Density of GaN (g/cm³)
Z_GaN = 31.5  # Average atomic number of GaN

def kanaya_okayama_depth(E, A, rho, Z):
    """
    Calculate penetration depth using Kanaya-Okayama equation
    
    Parameters:
    E: beam energy (keV)
    A: atomic mass (g/mol)
    rho: density (g/cm³)
    Z: atomic number
    
    Returns:
    R: penetration depth (μm)
    """
    R = 0.0276 * A * (E**1.67) / (rho * (Z**0.89))
    return R

# Calculate penetration depths for all beam energies
penetration_depths = []
for beam_energy in beam_energies:
    depth = kanaya_okayama_depth(beam_energy, A_GaN, rho_GaN, Z_GaN)
    penetration_depths.append(depth)

penetration_depths = np.array(penetration_depths)

# Find beam energies where QW1 and QW2 emissions are strongest
qw1_max_beam = beam_energies[np.argmax(depth_profiles['QW1'])]
qw2_max_beam = beam_energies[np.argmax(depth_profiles['QW2'])]

# Calculate corresponding penetration depths
qw1_depth = kanaya_okayama_depth(qw1_max_beam, A_GaN, rho_GaN, Z_GaN)
qw2_depth = kanaya_okayama_depth(qw2_max_beam, A_GaN, rho_GaN, Z_GaN)

print("Quantum Well Depth Estimation")
print("=" * 40)
print(f"Material properties assumed (GaN):")
print(f"  Atomic mass: {A_GaN} g/mol")
print(f"  Density: {rho_GaN} g/cm³")
print(f"  Atomic number: {Z_GaN}")
print()
print(f"QW1 Analysis:")
print(f"  Maximum intensity at: {qw1_max_beam} keV")
print(f"  Estimated penetration depth: {qw1_depth:.2f} μm")
print()
print(f"QW2 Analysis:")
print(f"  Maximum intensity at: {qw2_max_beam} keV")
print(f"  Estimated penetration depth: {qw2_depth:.2f} μm")
print()

# Create a plot showing penetration depth vs beam energy
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(12, 10))

# Plot 1: Penetration depth vs beam energy
ax1.plot(beam_energies, penetration_depths, 'b-', linewidth=3, marker='o', markersize=4)
ax1.axvline(x=qw1_max_beam, color='green', linestyle='--', alpha=0.7, linewidth=2, label=f'QW1 peak at {qw1_max_beam} keV')
ax1.axvline(x=qw2_max_beam, color='orange', linestyle='--', alpha=0.7, linewidth=2, label=f'QW2 peak at {qw2_max_beam} keV')
ax1.set_xlabel('Beam Energy (keV)', fontsize=12)
ax1.set_ylabel('Penetration Depth (μm)', fontsize=12)
ax1.set_title('Kanaya-Okayama Penetration Depth vs Beam Energy', fontsize=14, fontweight='bold')
ax1.grid(True, alpha=0.3)
ax1.legend()

# Plot 2: QW intensity profiles overlaid with penetration depth
ax2_twin = ax2.twinx()

# Plot QW intensities on left axis
ax2.plot(beam_energies, depth_profiles['QW1'], color='green', linewidth=3, marker='o', markersize=4, label='QW1 Intensity')
ax2.plot(beam_energies, depth_profiles['QW2'], color='orange', linewidth=3, marker='o', markersize=4, label='QW2 Intensity')
ax2.set_xlabel('Beam Energy (keV)', fontsize=12)
ax2.set_ylabel('CL Intensity', fontsize=12, color='black')
ax2.tick_params(axis='y', labelcolor='black')

# Plot penetration depth on right axis
ax2_twin.plot(beam_energies, penetration_depths, 'b--', linewidth=2, alpha=0.7, label='Penetration Depth')
ax2_twin.set_ylabel('Penetration Depth (μm)', fontsize=12, color='blue')
ax2_twin.tick_params(axis='y', labelcolor='blue')

ax2.set_title('QW Intensities vs Penetration Depth', fontsize=14, fontweight='bold')
ax2.grid(True, alpha=0.3)
ax2.legend(loc='upper left')
ax2_twin.legend(loc='upper right')

plt.tight_layout()
plt.show()
```

### Methodology and Interpretation for GaN LED Structure

**How the depth estimation works:**

1. **Identify Peak Beam Energies**: From the depth profiles, we identify the beam energies where each quantum well emission (QW1 and QW2) reaches maximum intensity.

2. **Apply Kanaya-Okayama Equation**: Calculate the maximum electron penetration depth for these specific beam energies using GaN material properties.

3. **Depth Correlation**: The beam energy that maximizes a particular QW emission corresponds approximately to the depth where that quantum well is located within the LED structure.

**Key Assumptions for GaN LED:**
- Material properties of GaN used throughout (though actual LED has varying composition)
- Maximum emission intensity correlates with optimal excitation depth
- Quantum wells are thin (~2-10 nm) compared to the interaction volume
- LED structure has p-type layers near surface, active region with QWs, then n-type layers

**Physical Reasoning for LED Structure:**
- **Low beam energies**: Mainly excite p-type surface layers, limited QW activation
- **Optimal beam energy**: Maximum electron density reaches the QW depth in active region
- **High beam energies**: Electrons penetrate through active region into n-type layers, reducing QW excitation efficiency

**LED-Specific Considerations:**
- The active region (containing QWs) is designed to be at an optimal depth for current injection
- p-n junction electric fields can affect carrier transport and recombination
- Different doping levels create varying recombination environments

**Limitations:**
- Kanaya-Okayama assumes uniform material, but LED has layered structure
- Peak generation occurs at ~1/3 of maximum penetration depth
- Electric fields in p-n junction not accounted for in simple penetration model

```python
# Refined depth estimation accounting for peak generation depth
# Peak carrier generation typically occurs at ~1/3 of maximum penetration depth

qw1_estimated_depth = qw1_depth / 3  # More accurate estimate for peak generation
qw2_estimated_depth = qw2_depth / 3

print("REFINED Quantum Well Depth Estimates")
print("=" * 45)
print("(Accounting for peak carrier generation at ~1/3 max penetration)")
print()
print(f"QW1 (2.634 eV emission):")
print(f"  Peak intensity beam energy: {qw1_max_beam} keV")
print(f"  Maximum penetration depth: {qw1_depth:.2f} μm")
print(f"  Estimated QW1 depth: {qw1_estimated_depth:.2f} μm ({qw1_estimated_depth*1000:.0f} nm)")
print()
print(f"QW2 (2.5 eV emission):")
print(f"  Peak intensity beam energy: {qw2_max_beam} keV")
print(f"  Maximum penetration depth: {qw2_depth:.2f} μm") 
print(f"  Estimated QW2 depth: {qw2_estimated_depth:.2f} μm ({qw2_estimated_depth*1000:.0f} nm)")
print()

# Determine which is deeper
if qw1_estimated_depth > qw2_estimated_depth:
    deeper_qw = "QW1"
    shallower_qw = "QW2"
    depth_difference = qw1_estimated_depth - qw2_estimated_depth
else:
    deeper_qw = "QW2"
    shallower_qw = "QW1"
    depth_difference = qw2_estimated_depth - qw1_estimated_depth

print(f"Structure Analysis:")
print(f"  Shallower QW: {shallower_qw}")
print(f"  Deeper QW: {deeper_qw}")
print(f"  Separation: {depth_difference:.2f} μm ({depth_difference*1000:.0f} nm)")

# Create a simple depth diagram
fig, ax = plt.subplots(figsize=(10, 6))

# Draw surface
ax.axhline(y=0, color='black', linewidth=3, label='Sample Surface')

# Draw quantum wells
qw1_y = -qw1_estimated_depth
qw2_y = -qw2_estimated_depth

ax.axhline(y=qw1_y, color='green', linewidth=4, alpha=0.8, label=f'QW1 at {qw1_estimated_depth:.2f} μm')
ax.axhline(y=qw2_y, color='orange', linewidth=4, alpha=0.8, label=f'QW2 at {qw2_estimated_depth:.2f} μm')

# Add annotations
ax.annotate(f'QW1\n{qw1_estimated_depth*1000:.0f} nm', 
           xy=(0.5, qw1_y), xytext=(0.7, qw1_y),
           fontsize=12, ha='left', va='center',
           arrowprops=dict(arrowstyle='->', color='green'))

ax.annotate(f'QW2\n{qw2_estimated_depth*1000:.0f} nm', 
           xy=(0.5, qw2_y), xytext=(0.7, qw2_y),
           fontsize=12, ha='left', va='center',
           arrowprops=dict(arrowstyle='->', color='orange'))

ax.set_xlim(0, 1)
ax.set_ylim(-max(qw1_estimated_depth, qw2_estimated_depth)*1.2, 0.1)
ax.set_ylabel('Depth (μm)', fontsize=14)
ax.set_title('Estimated Quantum Well Depths', fontsize=16, fontweight='bold')
ax.legend(fontsize=12)
ax.grid(True, alpha=0.3)

# Remove x-axis as it's not meaningful
ax.set_xticks([])

plt.tight_layout()
plt.show()
```

### 4.7 Critical Analysis: Physical Realism of NBE Depth Profile

#### 4.7.1 Expected vs Observed Behavior

The near-bandgap emission (NBE) depth profile exhibits behavior that requires careful scrutiny from a physical perspective. Our analysis reveals that NBE intensity continuously increases with beam energy, reaching maximum intensity at 30 keV. This behavior is **physically unrealistic** for a typical GaN LED structure for several reasons:

#### 4.7.2 Physical Expectations for NBE in GaN LEDs

1. **NBE Origin**: Near-bandgap emission (~3.4 eV) originates from direct band-to-band recombination in high-quality GaN material.

2. **Expected Depth Distribution**: NBE should be strongest near the surface and decrease with depth because:
   - Surface and near-surface regions typically have the highest crystal quality
   - Deeper regions, especially the n-type substrate, often contain higher defect densities
   - The active region is specifically designed for quantum well emission, not NBE

3. **Typical LED Architecture**: In optimized LED structures, the p-type surface layers are designed for current injection and light extraction, not for strong NBE generation.

#### 4.7.3 Problematic Aspects of Observed NBE Profile

1. **Continuous Increase**: The monotonic increase with beam energy contradicts the expected surface preference of high-quality GaN emission.

2. **Deep Maximum**: Peak intensity at 30 keV suggests strongest NBE from the deepest probed regions, which is inconsistent with typical LED design and material quality distributions.

3. **Absence of Surface Peak**: The profile shows minimal NBE at low beam energies, opposite to the expected behavior for surface-quality-dependent emission.

#### 4.7.4 Possible Explanations

Several factors could contribute to this unrealistic behavior:

1. **Spectral Overlap**: The 3.385 eV peak may overlap with other emissions or scattered light, leading to misidentification.

2. **Substrate Effects**: High beam energies may excite unexpected luminescence from the sapphire substrate or interfacial regions.

3. **Experimental Artifacts**: Potential issues include sample charging, beam damage, or systematic measurement errors at high beam energies.

4. **Peak Misassignment**: The feature identified as NBE may actually represent a different emission mechanism that becomes prominent at high beam energies.

#### 4.7.5 Implications for Data Interpretation

This analysis demonstrates the importance of critical evaluation when interpreting CL depth profiles. While the quantum well emissions (QW1, QW2) and blue luminescence (BL) show physically reasonable behavior consistent with LED structure expectations, the NBE profile suggests either experimental limitations or unusual sample properties that deviate from standard LED behavior.

## Method Comparison: Peak vs Threshold/Integration for CL Analysis

### Two Main Approaches for Extracting Emission Intensities:

1. **Peak Method (Current)**: Find maximum intensity within energy range
   - ✅ Simple and fast
   - ✅ Good for well-separated peaks
   - ❌ Sensitive to noise spikes
   - ❌ Ignores peak width/shape information

2. **Threshold/Integration Method**: Integrate area above background
   - ✅ More robust against noise
   - ✅ Accounts for peak broadening
   - ✅ Better represents total emission
   - ❌ Requires background subtraction
   - ❌ More complex implementation

### Which Method is Better for CL Depth Profiling?

For **cathodoluminescence depth profiling**, the **integration method** is generally preferred because:
- CL peaks can broaden with beam energy due to interaction volume effects
- Peak shape contains information about emission mechanisms
- Integration is less sensitive to experimental noise
- Total emission (area) is more physically meaningful than peak height

```python
# Compare Peak Method vs Integration Method
print("COMPARISON: PEAK METHOD vs INTEGRATION METHOD")
print("=" * 60)

def extract_integrated_intensity(energy_array, spectrum, peak_energy, energy_range, threshold_fraction=0.1):
    """
    Extract integrated intensity above background threshold
    
    Parameters:
    energy_array: photon energy values
    spectrum: intensity values
    peak_energy: center energy of emission band
    energy_range: energy range around peak
    threshold_fraction: fraction of peak height to use as threshold
    
    Returns:
    integrated_intensity: area above threshold
    """
    # Find indices within the energy range around the peak
    mask = (energy_array >= peak_energy - energy_range/2) & (energy_array <= peak_energy + energy_range/2)
    
    if not np.any(mask):
        return 0
    
    # Extract spectrum section
    energy_section = energy_array[mask]
    intensity_section = spectrum[mask]
    
    # Find peak height in this region
    peak_height = np.max(intensity_section)
    
    # Define threshold as fraction of peak height
    threshold = peak_height * threshold_fraction
    
    # Integrate area above threshold
    above_threshold = intensity_section > threshold
    if np.any(above_threshold):
        # Simple trapezoidal integration
        energy_step = np.mean(np.diff(energy_section))
        integrated_area = np.sum((intensity_section[above_threshold] - threshold)) * energy_step
        return integrated_area
    else:
        return 0

# Extract depth profiles using both methods
depth_profiles_peak = {}  # Current method (already calculated)
depth_profiles_integrated = {}  # New integration method

# Use the same emission bands as before
for band_name, band_info in emission_bands.items():
    # Peak method (already done)
    depth_profiles_peak[band_name] = depth_profiles[band_name]
    
    # Integration method
    integrated_intensities = []
    for i in range(30):  # For each beam energy (1-30 keV)
        spectrum = spectra[:, i]
        integrated_intensity = extract_integrated_intensity(energy, spectrum, 
                                                          band_info['center'], 
                                                          band_info['range'])
        integrated_intensities.append(integrated_intensity)
    
    depth_profiles_integrated[band_name] = np.array(integrated_intensities)

# Compare the two methods for each emission band
fig, axes = plt.subplots(2, 3, figsize=(18, 12))
axes = axes.flatten()

for i, (band_name, band_info) in enumerate(emission_bands.items()):
    ax = axes[i]
    
    # Normalize both profiles to 0-1 for comparison
    peak_profile = depth_profiles_peak[band_name]
    integrated_profile = depth_profiles_integrated[band_name]
    
    peak_norm = (peak_profile - np.min(peak_profile)) / (np.max(peak_profile) - np.min(peak_profile))
    integrated_norm = (integrated_profile - np.min(integrated_profile)) / (np.max(integrated_profile) - np.min(integrated_profile))
    
    ax.plot(beam_energies, peak_norm, 'o-', linewidth=2, markersize=4, 
           label='Peak Method', alpha=0.8)
    ax.plot(beam_energies, integrated_norm, 's-', linewidth=2, markersize=4, 
           label='Integration Method', alpha=0.8)
    
    ax.set_title(f'{band_name} ({band_info["center"]} eV)', fontweight='bold')
    ax.set_xlabel('Beam Energy (keV)')
    ax.set_ylabel('Normalized Intensity')
    ax.legend()
    ax.grid(True, alpha=0.3)
    
    # Find optimal beam energies for both methods
    peak_optimal = beam_energies[np.argmax(peak_profile)]
    integrated_optimal = beam_energies[np.argmax(integrated_profile)]
    
    ax.axvline(x=peak_optimal, color='blue', linestyle='--', alpha=0.7, linewidth=1)
    ax.axvline(x=integrated_optimal, color='orange', linestyle='--', alpha=0.7, linewidth=1)
    
    # Add text showing optimal beam energies
    ax.text(0.05, 0.95, f'Peak: {peak_optimal} keV', transform=ax.transAxes, 
           bbox=dict(boxstyle="round,pad=0.3", facecolor='lightblue', alpha=0.7), fontsize=9)
    ax.text(0.05, 0.85, f'Integrated: {integrated_optimal} keV', transform=ax.transAxes, 
           bbox=dict(boxstyle="round,pad=0.3", facecolor='lightyellow', alpha=0.7), fontsize=9)

# Remove empty subplot
axes[5].remove()

plt.tight_layout()
plt.show()

# Print comparison results
print("\nCOMPARISON RESULTS:")
print("=" * 50)
print(f"{'Band':<4} {'Peak Method':<12} {'Integration':<12} {'Difference':<12}")
print("-" * 50)

for band_name in emission_bands.keys():
    peak_optimal = beam_energies[np.argmax(depth_profiles_peak[band_name])]
    integrated_optimal = beam_energies[np.argmax(depth_profiles_integrated[band_name])]
    difference = abs(peak_optimal - integrated_optimal)
    
    print(f"{band_name:<4} {peak_optimal:<12} {integrated_optimal:<12} {difference:<12}")

print("\nKEY OBSERVATIONS:")
print("=" * 30)

# Check which method gives more consistent results
differences = []
for band_name in emission_bands.keys():
    peak_optimal = beam_energies[np.argmax(depth_profiles_peak[band_name])]
    integrated_optimal = beam_energies[np.argmax(depth_profiles_integrated[band_name])]
    differences.append(abs(peak_optimal - integrated_optimal))

mean_difference = np.mean(differences)
print(f"Average difference between methods: {mean_difference:.1f} keV")

if mean_difference < 2:
    print("✓ Both methods give similar results - peak method is sufficient")
elif mean_difference < 5:
    print("⚠ Moderate differences - integration method may be more reliable")
else:
    print("❌ Large differences - integration method strongly recommended")

# Analyze which bands show biggest differences
max_diff_band = list(emission_bands.keys())[np.argmax(differences)]
max_diff_value = np.max(differences)
print(f"Largest difference: {max_diff_band} ({max_diff_value} keV)")

# Check for noise sensitivity
print(f"\nNOISE SENSITIVITY ANALYSIS:")
for band_name in emission_bands.keys():
    peak_profile = depth_profiles_peak[band_name]
    integrated_profile = depth_profiles_integrated[band_name]
    
    # Calculate relative standard deviation (noise measure)
    peak_rsd = np.std(peak_profile) / np.mean(peak_profile)
    integrated_rsd = np.std(integrated_profile) / np.mean(integrated_profile)
    
    if integrated_rsd < peak_rsd:
        noise_verdict = "Integration less noisy ✓"
    else:
        noise_verdict = "Peak less noisy"
    
    print(f"{band_name}: Peak RSD={peak_rsd:.2f}, Integrated RSD={integrated_rsd:.2f} - {noise_verdict}")

# Final recommendation
print(f"\n{'='*60}")
print("RECOMMENDATION:")
print(f"{'='*60}")

if mean_difference < 2 and np.mean([np.std(depth_profiles_peak[band])/np.mean(depth_profiles_peak[band]) 
                                   for band in emission_bands.keys()]) < 0.5:
    print("✓ PEAK METHOD is sufficient for this dataset")
    print("  - Results are consistent between methods")
    print("  - Peak method is simpler and faster")
    print("  - Good signal-to-noise ratio")
else:
    print("✓ INTEGRATION METHOD is recommended")
    print("  - More robust against noise")
    print("  - Better represents total emission")
    print("  - More physically meaningful for CL analysis")
```

```python
# IMPROVED Method Comparison: Peak vs Integration
print("IMPROVED COMPARISON: PEAK vs INTEGRATION METHODS")
print("=" * 60)

def extract_integrated_intensity_improved(energy_array, spectrum, peak_energy, energy_range):
    """
    Improved integration method: integrate peak area above baseline
    """
    # Find indices within the energy range around the peak
    mask = (energy_array >= peak_energy - energy_range/2) & (energy_array <= peak_energy + energy_range/2)
    
    if not np.any(mask):
        return 0
    
    # Extract spectrum section
    energy_section = energy_array[mask]
    intensity_section = spectrum[mask]
    
    if len(intensity_section) < 3:
        return np.max(intensity_section)  # Fall back to peak if too few points
    
    # Estimate baseline using edge points
    baseline = (intensity_section[0] + intensity_section[-1]) / 2
    
    # Subtract baseline and integrate positive area
    corrected_intensity = intensity_section - baseline
    corrected_intensity[corrected_intensity < 0] = 0  # Only positive contributions
    
    # Simple trapezoidal integration
    energy_step = np.mean(np.diff(energy_section)) if len(energy_section) > 1 else 0.01
    integrated_area = np.trapz(corrected_intensity, dx=energy_step)
    
    return max(integrated_area, 0)

# Extract depth profiles using improved integration method
depth_profiles_integrated_v2 = {}

for band_name, band_info in emission_bands.items():
    integrated_intensities = []
    for i in range(30):  # For each beam energy (1-30 keV)
        spectrum = spectra[:, i]
        integrated_intensity = extract_integrated_intensity_improved(energy, spectrum, 
                                                                   band_info['center'], 
                                                                   band_info['range'])
        integrated_intensities.append(integrated_intensity)
    
    depth_profiles_integrated_v2[band_name] = np.array(integrated_intensities)

# Create detailed comparison for key emission bands
fig, axes = plt.subplots(2, 2, figsize=(15, 10))
axes = axes.flatten()

key_bands = ['BL', 'QW1', 'QW2', 'NBE']  # Focus on most important bands

for i, band_name in enumerate(key_bands):
    ax = axes[i]
    
    # Get profiles
    peak_profile = depth_profiles_peak[band_name]
    integrated_profile = depth_profiles_integrated_v2[band_name]
    
    # Plot both methods (raw values, not normalized)
    ax2 = ax.twinx()
    
    # Peak method (left axis)
    line1 = ax.plot(beam_energies, peak_profile, 'o-', linewidth=2, markersize=4, 
                   color='blue', label='Peak Method', alpha=0.8)
    ax.set_ylabel('Peak Intensity', color='blue', fontsize=11)
    ax.tick_params(axis='y', labelcolor='blue')
    
    # Integration method (right axis)
    line2 = ax2.plot(beam_energies, integrated_profile, 's-', linewidth=2, markersize=4, 
                    color='orange', label='Integration Method', alpha=0.8)
    ax2.set_ylabel('Integrated Intensity', color='orange', fontsize=11)
    ax2.tick_params(axis='y', labelcolor='orange')
    
    # Find optimal beam energies
    peak_optimal = beam_energies[np.argmax(peak_profile)]
    integrated_optimal = beam_energies[np.argmax(integrated_profile)]
    
    # Mark optimal points
    ax.axvline(x=peak_optimal, color='blue', linestyle='--', alpha=0.7, linewidth=2)
    ax2.axvline(x=integrated_optimal, color='orange', linestyle='--', alpha=0.7, linewidth=2)
    
    ax.set_title(f'{band_name} ({emission_bands[band_name]["center"]} eV)', fontweight='bold', fontsize=12)
    ax.set_xlabel('Beam Energy (keV)', fontsize=11)
    ax.grid(True, alpha=0.3)
    
    # Add legend
    lines = line1 + line2
    labels = [l.get_label() for l in lines]
    ax.legend(lines, labels, loc='upper left', fontsize=10)
    
    # Add text showing optimal beam energies
    ax.text(0.02, 0.98, f'Peak: {peak_optimal} keV', transform=ax.transAxes, 
           bbox=dict(boxstyle="round,pad=0.3", facecolor='lightblue', alpha=0.8), fontsize=10,
           verticalalignment='top')
    ax.text(0.02, 0.85, f'Integrated: {integrated_optimal} keV', transform=ax.transAxes, 
           bbox=dict(boxstyle="round,pad=0.3", facecolor='moccasin', alpha=0.8), fontsize=10,
           verticalalignment='top')

plt.tight_layout()
plt.show()

# Print detailed comparison
print("\nDETAILED COMPARISON RESULTS:")
print("=" * 50)
print(f"{'Band':<4} {'Peak Optimal':<12} {'Integrated Optimal':<18} {'Difference':<10} {'Agreement':<12}")
print("-" * 70)

agreements = []
for band_name in emission_bands.keys():
    peak_optimal = beam_energies[np.argmax(depth_profiles_peak[band_name])]
    integrated_optimal = beam_energies[np.argmax(depth_profiles_integrated_v2[band_name])]
    difference = abs(peak_optimal - integrated_optimal)
    
    if difference <= 1:
        agreement = "Excellent"
    elif difference <= 3:
        agreement = "Good"
    elif difference <= 5:
        agreement = "Moderate"
    else:
        agreement = "Poor"
    
    agreements.append(agreement)
    print(f"{band_name:<4} {peak_optimal:<12} {integrated_optimal:<18} {difference:<10} {agreement:<12}")

# Analysis of which method is more reliable
print(f"\n{'='*60}")
print("METHOD RELIABILITY ANALYSIS:")
print(f"{'='*60}")

# Check correlation between methods
correlations = []
for band_name in emission_bands.keys():
    peak_profile = depth_profiles_peak[band_name]
    integrated_profile = depth_profiles_integrated_v2[band_name]
    
    # Calculate correlation coefficient
    correlation = np.corrcoef(peak_profile, integrated_profile)[0, 1]
    correlations.append(correlation)
    
    print(f"{band_name}: Correlation = {correlation:.3f}")

mean_correlation = np.mean(correlations)
print(f"\nAverage correlation between methods: {mean_correlation:.3f}")

# Check which bands have good agreement
good_agreement = [band for band, agreement in zip(emission_bands.keys(), agreements) 
                 if agreement in ['Excellent', 'Good']]
poor_agreement = [band for band, agreement in zip(emission_bands.keys(), agreements) 
                 if agreement in ['Poor']]

print(f"\nBands with good agreement: {', '.join(good_agreement)}")
if poor_agreement:
    print(f"Bands with poor agreement: {', '.join(poor_agreement)}")

# Final recommendation based on analysis
print(f"\n{'='*60}")
print("FINAL RECOMMENDATION:")
print(f"{'='*60}")

if mean_correlation > 0.8 and len(good_agreement) >= 3:
    print("✓ PEAK METHOD is adequate for this dataset")
    print("  - High correlation between methods")
    print("  - Good agreement for most emission bands")
    print("  - Peak method is simpler and computationally efficient")
    print("  - Suitable for well-separated peaks with good signal-to-noise")
else:
    print("✓ INTEGRATION METHOD is recommended")
    print("  - Provides more robust quantification")
    print("  - Less sensitive to peak shape variations")
    print("  - Better for overlapping or broad peaks")

print(f"\nFor this GaN LED analysis:")
if 'BL' in good_agreement and 'QW2' in good_agreement:
    print("- Peak method works well for BL and QW emissions (well-separated peaks)")
if 'NBE' in poor_agreement or 'YL' in poor_agreement:
    print("- Integration method recommended for NBE/YL (potential overlap issues)")

print(f"\n⚠️ Note: The choice also depends on your specific research goals:")
print("- Peak method: Quick screening and relative comparisons")
print("- Integration method: Quantitative analysis and publication-quality results")
```

## 5. Conclusions

This comprehensive depth-resolved cathodoluminescence study of a GaN-based LED structure has successfully demonstrated the power of variable-energy CL spectroscopy for characterizing complex semiconductor devices. The key findings and conclusions are:

### 5.1 Structural Characterization

The analysis identified five distinct emission bands and revealed the internal structure of the LED device:

- **Quantum Well Positions**: QW2 is located at ~350 nm depth (optimal at 13 keV), while QW1 is positioned deeper at ~720 nm (optimal at 20 keV), demonstrating a multi-quantum-well architecture designed for enhanced light extraction and current distribution.

- **Blue Luminescence Distribution**: BL emission shows realistic surface-preferential behavior with maximum intensity at 8 keV, consistent with defect or dopant-related centers concentrated in the near-surface p-type regions.

### 5.2 Methodological Insights

- **Peak Extraction Validity**: The peak height method proved effective for well-separated emission bands, providing reliable depth profiles for QW1, QW2, and BL emissions. This approach is suitable for initial analysis and comparative studies of LED structures.

- **Physical Realism Assessment**: Critical evaluation of depth profiles is essential. While most emissions showed physically reasonable behavior, the NBE profile exhibited unrealistic characteristics, highlighting potential experimental artifacts or measurement limitations.

### 5.3 LED Design Implications

The results provide insights into the LED structure optimization:
- The quantum wells are positioned at different depths, suggesting a design strategy to optimize current injection and light extraction efficiency
- The separation of ~370 nm between QW1 and QW2 indicates sophisticated epitaxial growth control
- The surface preference of BL emission suggests effective p-type doping near the contact region

### 5.4 Technical Recommendations

For future CL depth profiling studies of LED structures:

1. **Multiple Analysis Methods**: Combine peak height extraction with integration methods for overlapping or broad peaks
2. **Critical Evaluation**: Always assess physical realism of depth profiles, particularly for emissions that show unexpected behavior
3. **Complementary Techniques**: Consider supporting measurements (TEM, SIMS) to validate depth estimates and structural assignments

### 5.5 Broader Impact

This work demonstrates that variable-energy CL spectroscopy is a valuable non-destructive technique for characterizing buried quantum well structures in commercial LED devices. The methodology can be extended to other semiconductor devices and provides a foundation for quality control and failure analysis in LED manufacturing.

The successful identification of quantum well positions and the critical analysis of potentially problematic emission profiles showcase both the capabilities and limitations of the technique, providing a realistic assessment that will benefit researchers working on similar semiconductor characterization challenges.
